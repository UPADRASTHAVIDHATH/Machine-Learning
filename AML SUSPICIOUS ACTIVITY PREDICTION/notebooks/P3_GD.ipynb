## 1. Gradient Descent Implementation


```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -250, 250)))

def compute_loss(X, y, weights):
    m = len(y)
    h = sigmoid(X.dot(weights))
    return - (1/m) * np.sum(y * np.log(h + 1e-15) + (1 - y) * np.log(1 - h + 1e-15))

print("Gradient Descent functions compiled.")
```
