## 1. Random Forest with OOB Evaluation


```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

X_train = pd.read_csv('../data/X_train_processed.csv')
y_train = pd.read_csv('../data/y_train_encoded.csv').values.ravel()

rf_oob = RandomForestClassifier(n_estimators=100, oob_score=True, class_weight='balanced', random_state=42)
rf_oob.fit(X_train, y_train)
print(f"OOB Accuracy Score on AML Dataset: {rf_oob.oob_score_:.4f}")
```
