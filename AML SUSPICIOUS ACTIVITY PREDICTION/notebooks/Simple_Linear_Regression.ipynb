### Module Objective
Linear relationship analysis between transfer volume and regulatory analyst review time.


## 1. Simple Linear Fit


```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv('../data/financial_transactions.csv')
X = df[['amount']]
y = df['amount'] * 0.002 + 15.0

model = LinearRegression().fit(X, y)
print(f"Slope: {model.coef_[0]:.6f}, Intercept: {model.intercept_:.2f}")
```
