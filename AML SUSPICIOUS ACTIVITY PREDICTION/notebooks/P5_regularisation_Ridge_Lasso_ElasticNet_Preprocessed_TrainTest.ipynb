## 1. Regularized Models Training


```python
import pandas as pd
from sklearn.linear_model import RidgeClassifier, SGDClassifier

X_train = pd.read_csv('../data/X_train_processed.csv')
y_train = pd.read_csv('../data/y_train_encoded.csv').values.ravel()

ridge = RidgeClassifier(alpha=1.0)
ridge.fit(X_train, y_train)

elastic = SGDClassifier(loss='log_loss', penalty='elasticnet', l1_ratio=0.5, random_state=42)
elastic.fit(X_train, y_train)
print("Ridge and ElasticNet models trained successfully.")
```
