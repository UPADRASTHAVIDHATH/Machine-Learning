### Module Objective
Modeling continuous laundering risk index as a linear combination of financial velocity, account drain, and jurisdiction index.


## 1. Multiple Regression Implementation


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

df = pd.read_csv('../data/DT_AML_Transactions.csv')
features = ['amount', 'customerRiskScore', 'countryRiskIndex', 'transactionVelocity']
X = df[features]
y = df['customerRiskScore'] * 0.5 + df['transactionVelocity'] * 4.2

lr = LinearRegression()
lr.fit(X, y)
preds = lr.predict(X)
print("Coefficients:", lr.coef_)
print(f"R2 Score: {r2_score(y, preds):.4f}")
```
