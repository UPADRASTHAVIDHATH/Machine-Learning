### Module Objective
Fitting a binary classification model to predict legitimate vs illicit laundering transactions.


## 1. Logistic Regression Training


```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X_train = pd.read_csv('../data/X_train_processed.csv')
X_test = pd.read_csv('../data/X_test_processed.csv')
y_train = pd.read_csv('../data/y_train_encoded.csv').values.ravel()
y_test = pd.read_csv('../data/y_test_encoded.csv').values.ravel()

clf = LogisticRegression(class_weight='balanced', max_iter=1000)
clf.fit(X_train, y_train)
preds = clf.predict(X_test)
print(classification_report(y_test, preds, zero_division=0))
```
