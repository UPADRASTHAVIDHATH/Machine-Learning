### Module Objective
Multiclass logistic classification predicting risk levels: Tier 1 (Low), Tier 2 (Medium), Tier 3 (High), Tier 4 (Critical).


## 1. Multi-Class Risk Tier Classification


```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression

X_train = pd.read_csv('../data/X_train_processed.csv')
# Synthesize 4 risk tiers
y_multiclass = np.random.choice([0, 1, 2, 3], size=len(X_train), p=[0.7, 0.15, 0.1, 0.05])

multi_clf = LogisticRegression(multi_class='multinomial', max_iter=1000)
multi_clf.fit(X_train, y_multiclass)
print("Multinomial model fitted successfully across 4 AML Risk Tiers.")
```
