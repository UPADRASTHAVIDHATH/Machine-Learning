### Executive Summary
Optimizing decision thresholds and hyperparameters for Random Forest and Boosting models to maximize detection recall while penalizing false positives.


## 1. Setup Grid Search Parameters


```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

X_train = pd.read_csv('../dataset/X_train_processed.csv')
y_train = pd.read_csv('../dataset/y_train.csv').values.ravel()

param_grid = {
    'n_estimators': [100, 150],
    'max_depth': [10, 15],
    'min_samples_split': [2, 5],
    'class_weight': ['balanced', 'balanced_subsample']
}

rf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=3, scoring='f1', n_jobs=-1, verbose=1)
print("GridSearchCV configuration prepared.")
```

## 2. Execute Tuning and Inspect Best Parameters


```python
grid_search.fit(X_train, y_train)
print(f"Best Parameters: {grid_search.best_params_}")
print(f"Best Cross-Validation F1-Score: {grid_search.best_score_:.4f}")
```
