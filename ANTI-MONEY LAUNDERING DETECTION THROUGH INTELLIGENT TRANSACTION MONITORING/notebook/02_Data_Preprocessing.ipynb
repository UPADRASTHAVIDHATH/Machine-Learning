### Executive Summary
Pipeline for sanitizing financial transaction streams, log transformations for heavy-tailed payment amounts, and missing value verification.


## 1. Import Dependencies


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

print("Preprocessing pipeline initialized.")
```

## 2. Ingestion & Quality Checks


```python
df = pd.read_csv('../dataset/AML_Complete_Transaction_Dataset.csv')
print(f"Missing values count per column:
{df.isnull().sum()}")
```

## 3. Balance Consistency Checks


```python
# Validate mathematical consistency of ledger balances
df['errorBalanceOrig'] = df['newbalanceOrig'] + df['amount'] - df['oldbalanceOrg']
df['errorBalanceDest'] = df['oldbalanceDest'] + df['amount'] - df['newbalanceDest']

print(f"Average Originator Balance Error: ${df['errorBalanceOrig'].mean():.2f}")
print(f"Average Beneficiary Balance Error: ${df['errorBalanceDest'].mean():.2f}")
```

## 4. Log Transformation on Skewed Financial Values


```python
df['log_amount'] = np.log1p(df['amount'])
print("Skewness before transformation:", df['amount'].skew())
print("Skewness after log transformation:", df['log_amount'].skew())
```
