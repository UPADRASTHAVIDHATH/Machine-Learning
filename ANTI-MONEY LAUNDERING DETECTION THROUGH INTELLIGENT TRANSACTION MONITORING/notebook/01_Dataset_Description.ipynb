## 1. Environment & Library Setup


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Libraries loaded successfully.")
```

## 2. Load AML Transaction Dataset


```python
df = pd.read_csv('../dataset/AML_Complete_Transaction_Dataset.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 3. Statistical Summary and Schema Audit


```python
print("--- Column Data Types & Null Count ---")
print(df.info())
print("
--- Summary Statistics ---")
print(df.describe())
```

## 4. Class Distribution & Laundering Incident Rate


```python
laundering_counts = df['isLaundering'].value_counts()
laundering_pct = df['isLaundering'].value_counts(normalize=True) * 100

print(f"Legitimate Transactions: {laundering_counts.get(0, 0)} ({laundering_pct.get(0, 0):.2f}%)")
print(f"Illicit Laundering Transactions: {laundering_counts.get(1, 0)} ({laundering_pct.get(1, 0):.2f}%)")

plt.figure(figsize=(6, 4))
sns.countplot(x='isLaundering', data=df, palette=['#3b82f6', '#ef4444'])
plt.title("Transaction Class Distribution (0=Legitimate, 1=Money Laundering)")
plt.xlabel("Laundering Flag")
plt.ylabel("Transaction Count")
plt.show()
```
