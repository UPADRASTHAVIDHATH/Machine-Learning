### Executive Summary
Comparative benchmarking of baseline models versus the final optimized ensemble across Accuracy, Precision, Recall, F1, and ROC-AUC.


## 1. Load Benchmark Results


```python
import pandas as pd
import matplotlib.pyplot as plt

df_comp = pd.read_csv('model_comparison_final.csv')
print("Model Comparison Benchmark Table:")
display(df_comp) if 'display' in locals() else print(df_comp)
```

## 2. Benchmark Visualization


```python
df_plot = df_comp.set_index('Model')[['Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC']]
df_plot.plot(kind='bar', figsize=(12, 6), colormap='viridis')
plt.title("Algorithm Comparison Across AML Performance Metrics")
plt.ylabel("Score")
plt.xticks(rotation=25, ha='right')
plt.ylim(0.7, 1.05)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
