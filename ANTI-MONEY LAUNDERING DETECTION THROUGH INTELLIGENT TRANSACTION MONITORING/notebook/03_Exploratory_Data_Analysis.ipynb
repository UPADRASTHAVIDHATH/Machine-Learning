### Executive Summary
In-depth exploratory analysis uncovering patterns in transaction velocity, structuring (smurfing), high-risk jurisdictions, and transaction modes.


## 1. Load Dataset & Visual Setup


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../dataset/AML_Complete_Transaction_Dataset.csv')
df.head(3)
```

## 2. Transaction Type Breakdown by Laundering Status


```python
plt.figure(figsize=(10, 5))
sns.countplot(x='type', hue='isLaundering', data=df, palette='coolwarm')
plt.title("Laundering Incidence Across Transaction Types")
plt.xlabel("Transaction Method")
plt.ylabel("Volume")
plt.legend(title="Is Laundering", labels=["Normal", "Illicit"])
plt.show()
```

## 3. Structuring / Smurfing Pattern Analysis


```python
structuring_summary = df.groupby('isStructuring')['isLaundering'].mean() * 100
print("Percentage of laundering in structured transactions ($9,000 - $9,999):")
print(structuring_summary)

plt.figure(figsize=(8, 4))
sns.barplot(x=structuring_summary.index, y=structuring_summary.values, palette='viridis')
plt.title("Laundering Probability: Regular vs Structured Amounts")
plt.ylabel("Laundering Probability (%)")
plt.xlabel("Is Structuring (<$10k Cash Band)")
plt.show()
```

## 4. Correlation Heatmap


```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, cmap='vlag', annot=False, linewidths=0.5)
plt.title("Correlation Matrix of Financial & AML Risk Features")
plt.show()
```
